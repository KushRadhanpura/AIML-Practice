**Problem-1** salary prediction



```python
# simple linear regression for salary prediction
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# dummy data - X must be 2d array else sklearn throws error
X = np.array([[1.1], [1.5], [2.0], [2.5], [3.0]])  # exp
y = np.array([40000, 45000, 52000, 60000, 65000])  # salary
```

```python
# training
model = LinearRegression()
model.fit(X, y)
```

```python
# checking parameters (y = mx + c)
# note: trailing underscore means values are learned after fitting
m = model.coef_[0]
c = model.intercept_
print(f"equation: y = {m:.2f}*x + {c:.2f}")
```

```python
# test prediction
# need double brackets here to avoid valueerror shape issue
test_exp = [[2.8]]
pred = model.predict(test_exp)
print(f"pred for 2.8 yrs: {pred[0]:.2f}")
```

```python
# evaluation
y_pred = model.predict(X)
print("mse:", mean_squared_error(y, y_pred))
print("r2:", r2_score(y, y_pred))
```

**Problem-2** Telecom company


```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
#data setup(X=age,monthly bill y=Churn(stay=0 ,1=leave)
X=np.array([[21,50],[45,110],[30,60],[55,120],[23,45]])
y=np.array([0,1,0,1,0])
```

```python
#Train classification engine
model=LogisticRegression()
model.fit(X,y)
```

```python
#live prediction for new customer(age=40,monthly bill=95)
new_user=[[40,95]]
pred=model.predict(new_user)
print=(f"churn prediction for {new_user[0]}:{pred[0]}")
```

```python
#Evaluation matrix
y_pred=model.predict(X)
print("Accuracy score:",accuracy_score(y,y_pred))
print("Confusion matrix: ",confusion_matrix(y,y_pred))
```

**Problem-3** Data preprocessing

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
#Raw messy data set
raw_data={
    'Experience':[2.5,np.nan,5.0,1.2,np.nan,4.0],
    'Department':['IT','HR','IT','Sales',np.nan,'HR'],
    'Salary':[50000,35000,42000,70000,44000,66000]
}
```

```python
df=pd.DataFrame(raw_data)
print("Before preprocessing (messy data) ",df)
```

```python
#Handle Missing Numericals Values using Mean
num_imputer=SimpleImputer(strategy='mean')
df['Experience']=num_imputer.fit_transform(df[['Experience']])
```

```python
#Handle Missing Categorical Values using Mode Frequent(Mode)
cat_imputer=SimpleImputer(strategy='most_frequent')
df['Department']=cat_imputer.fit_transform(df[['Department']]).ravel()
```

```python
#One-Hot Encodeing for Text categories(Sales,IT,HR)
encoder=OneHotEncoder(sparse_output=False)
encoder_dept=encoder.fit_transform(df[['Department']])
```

```python
print("After Imputation(Numerical cleaned)",df ['Experience'].values)
print("One-Hot Encoded Department",encoder_dept)
```

**Problem 4** K-mwans clustering

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#Customer profile(Annual Income in lakhs,spending score 1-100 )
X=np.array([[15,39],[15,81],[19,6],[40,72],[47,40],[54,55],[78,17],[88,91]])
```

```python
#Configure KMeans to form 3 distinct groups
#Note:n_init='auto' ensures modern scikit-learn compliance
kmeans=KMeans(n_clusters=3,random_state=42,n_init='auto')
y_clusters=kmeans.fit_predict(X)
```

```python
print("Customer Group Assignment: ",y_clusters)
print("Calculate CLusters Centers: ",kmeans.cluster_centers_)
```

**Problem 5** Loan approval System (Decision tree classifier)


```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import export_text
```

```python
#X=[Credit score,income in lakhs],y=Approved(1=yes,2=no)
X=np.array([[600,5],[750,12],[500,3],[710,8],[680,4]])
y=np.array([0,1,0,1,0])
```

```python
#Train decision tree
tree_model=DecisionTreeClassifier(criterion='entropy',random_state=42)
tree_model.fit(X,y)
```

```python
#text output of logic tree splits
tree_rules=export_text(tree_model,feature_names=['credit_score','Income'])
print("Internal Decision tree rules",tree_rules)
```
