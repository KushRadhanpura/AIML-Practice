**Tokenizarion**

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
text="I love machine learning"
tokens =word_tokenize(text)
print(tokens)
```

**Stopwords Removal**

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize


text="My name is kush Radhanpura I am 3 year btech cse student Currently preparing for Intenship"
stop_words=set(stopwords.words('english'))
tokens=word_tokenize(text)
filtered=[word for word in tokens if word.lower() not in stop_words]
print(filtered)
```

**Stemming**

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()

words=['playing',"played","plays"]
for word in words:
  print(stemmer.stem(word))
```

**Lemmetization**

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
print(lemmatizer.lemmatize("running",pos="v"))
```

**Bag of Words**

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus=[
    "I love Anime",
    "I love Naruto",
    "I watch JJK"
]

vectorize=CountVectorizer()
X=vectorize.fit_transform(corpus)
print(vectorize.get_feature_names_out())
print(X.toarray())
```

**TF-IDF**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
corpus=[
    "I want internship in AI-ML",
    "I am begginer but learning the things"
]

vectorize=TfidfVectorizer()
X=vectorize.fit_transform(corpus)
print(vectorize.get_feature_names_out())
print(X.toarray())
```

**N-Grams**

```python
from sklearn.feature_extraction.text import CountVectorizer

text=["I love anime"]

vectorize=CountVectorizer(ngram_range=(1,2))
X=vectorize.fit_transform(text)
print(vectorize.get_feature_names_out())
```
